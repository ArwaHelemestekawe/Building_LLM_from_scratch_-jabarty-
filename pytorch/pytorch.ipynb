```python
w = torch.tensor([2.0],requires_grad=True)
x = torch.tensor([3.0])

loss=(w*x-1)**2
loss.backward()

```

```python
import torch
import torch.nn as nn

class Tiny_model(nn.Module):
  def __init__(self):
    super().__init__()
    #self.w=torch.tensor([2,],requires_grad=True)
    #self.b=torch.tensor([2,],requires_grad=True)
    self.w=nn.Parameter(torch.tensor([2.0]))
    self.b=nn.Parameter(torch.tensor([2.0]))

  def forward(self,x):
    return (self.w*x)+(self.b)

```

```python
my_model=Tiny_model()
my_input=torch.tensor([1.0])
print("w:",my_model.w)
print("b:",my_model.b)
print(my_model(my_input))
prediction=my_model(my_input)
print("my prediction",prediction)
```

```python
target=8
learning_rate=0.01
loss=(prediction-target)**2
loss
```

```python
loss.backward()
```

```python
print("w:",my_model.w)
print("b:",my_model.b)
print(my_model(my_input))
```

```python
print("w:",my_model.w.grad)
print("b:",my_model.b.grad)
```

```python
for step in range(10):
  prediction=my_model(my_input)
  loss=(prediction-target)**2
  loss.backward()
  print("current step :",step)
  print("prediction",prediction)
  print("loss:",loss)
  print("w:",my_model.w.grad)
  print("b:",my_model.b.grad)
  print("="*30)
  with torch.no_grad():
    # Use in-place subtraction (.sub_()) to update the parameter values directly
    my_model.w.sub_(learning_rate * my_model.w.grad)
    my_model.b.sub_(learning_rate * my_model.b.grad)

  # Correctly call the .zero_() function with parentheses to reset gradients
  my_model.w.grad.zero_()
  my_model.b.grad.zero_()
```

# Simple Neural Network Pipeline

## Steps

1. Take **input token IDs**.
2. Convert each token ID into a dense vector using an **Embedding Layer**.
3. Apply **Normalization** to the embeddings.
4. Pass the normalized embeddings through a **Feed-Forward Network**.
5. Generate a **score (logit)** for each token in the vocabulary.
6. Use the scores to predict the **next token**.

# Architecture Flow

## Input IDs
↓

## Embedding Layer
Convert token IDs into embedding vectors.

↓

## Normalization
Normalize the embedding values.

↓

## Feed-Forward Network
Process the embeddings and learn patterns.

↓

## Token Scores (Logits)
Generate a score for every token in the vocabulary.

↓

## Predicted Token
Select the token with the highest score.

```python
vocab_size=24000
dim_embedding=512
```

```python
import torch.nn as nn
class Simple (nn.Module):
  def __init__(self,vocab_size:int,dim_embedding:int) :
    self.embedding=nn.Embedding(vocab_size,dim_embedding)
    self.norm=nn.LayerNorm(dim_embedding)
    self.linear=nn.Linear(dim_embedding,vocab_size)
    self.softmax=nn.Softmax(dim=-1)

    super().__init__()
    pass

  def forward(self,x_input):
    embed_out=self.embedding(x_input)
    norm_out=self.norm(embed_out)
    linear_out=self.linear(norm_out)
    #Represent the model's score for each vocabulary token logits
    return self.softmax(linear_out)
```

```python
import torch
import torch.nn as nn

# Redefining Simple with super().__init__() at the beginning of __init__
class Simple(nn.Module):
  def __init__(self, vocab_size: int, dim_embedding: int):
    super().__init__() # Must be called first!
    self.embedding = nn.Embedding(vocab_size, dim_embedding)
    self.norm = nn.LayerNorm(dim_embedding)
    self.linear = nn.Linear(dim_embedding, vocab_size)
    self.softmax = nn.Softmax(dim=-1)

  def forward(self, x_input):
    embed_out = self.embedding(x_input)
    norm_out = self.norm(embed_out)
    linear_out = self.linear(norm_out)
    return self.softmax(linear_out)

simple = Simple(vocab_size=vocab_size,
                dim_embedding=dim_embedding)

tokens_id = torch.tensor(
    [
        [ 1, 2, 3],
        [ 1, 2, 3]
    ]
)

tokens = simple(tokens_id) # Using simple(tokens_id) is preferred over simple.forward(tokens_id)
print(tokens)
```
