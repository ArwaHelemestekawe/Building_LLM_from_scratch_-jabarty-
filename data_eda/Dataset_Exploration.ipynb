```python
import pandas as pd
from datasets import load_dataset
repo="bakrianoo/jabarti-llm-dataset"

pretrain=load_dataset(
    repo,
    "pretrain",split="phase1_train"

).to_pandas()

fine_tune=load_dataset(
    repo,
    "finetune",split="train"

).to_pandas()
```

```python
print("===pretrain schema====")
print(pretrain.dtypes)
print("===finetune schema====")
print(fine_tune.dtypes)
```

```python
splits={
    "pretrain / phase1_train":("pretrain","phase1_train"),
     "pretrain / phase1_eval":("pretrain","phase1_eval"),
    "pretrain / phase2_train":("pretrain","phase2_train"),
    "pretrain / phase2_eval":("pretrain","phase2_eval"),


    "finetune / train":("finetune","train"),
    "finetune / eval":("finetune","eval"),

}
n_char_token=3.5
context_length=[512,1024,2048]
```

```python
for name, (sub_set, split_name) in splits.items():
  df = load_dataset(repo, sub_set, split=split_name).to_pandas()

  print(f"data name: {name} | size: {df.shape[0]}")

  nulls = df.isnull().sum()
  nulls_filtered = nulls[nulls > 0]

  if nulls_filtered.empty:
      print("non_nulls: None")
  else:
      print("non_nulls:")
      print(nulls_filtered.to_string())
  print("-" * 30)
  total_row=df.shape[0]

  lang_count=df["language"].value_counts()
  for language ,n in lang_count.items():
      print(f"{language}: {n/total_row*100:.1f}%")
      print("-"*130)

  if sub_set=="pretrain":
    len_text=df["text"].str.len().describe()
    print(f"pretrain  len text:\t{len_text}")
  elif sub_set=="finetune":
    len_text=pd.concat([df["answer"].str.len(),df["question"].str.len()]).describe()
    print(f"finetune len text:\t{len_text}")

  print(f"{'='*30} tokens for {name} {'='*30}")

  if sub_set=="pretrain":
    sample_chars=df["text"].str.len()
  elif sub_set=="finetune":
    sample_chars=(
        df["question"].str.len()+
        df["answer"].str.len()
    )

  approximite_tokens=(
      sample_chars/n_char_token
  ).astype(int)

  print(approximite_tokens.describe().round(1))
  for ctx in context_length:
    dropped_pct=(
        (approximite_tokens>ctx).mean()*100)
    keep=100-dropped_pct
    print(f"context limit {ctx} -> dropped: {dropped_pct:.2f}%, kept: {keep:.2f}%")
  print("\n" + "="*80 + "\n")
```
